```python
import sympy as sy
import numpy as np
```

# Define symbols and basic matrices

```python
h, p, hb,pb,tb,l,r,g,m,c = sy.symbols('h,P,hb,Pb,Tb,L,R,g,M,c')

f1 = hb+tb/l * (1-(p/pb)**(c))
f2 = p
f3 = pb
f4 = hb
f5 = tb

f_x_u = sy.matrices.Matrix([f1,f2,f3,f4,f5])

x_k = sy.matrices.Matrix([h,p,pb,hb,tb])

sy.pprint(f_x_u)
sy.pprint(x_k)
```

# Compute the derivatives

```python
test = sy.Derivative(f_x_u,x_k,evaluate=True)

sy.pprint(test)
```

```python
x,y = sy.symbols("x,y")

sy.Derivative(x**2+y**2,x,evaluate=True)
```
